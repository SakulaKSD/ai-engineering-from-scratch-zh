# 3.git操作

工作目录→暂存区：git add<br><br>暂存区→本地仓库：git commit<br><br>本地仓库→远端（github）：git push<br><br>远端→本地仓库：git fetch<br><br>本地仓库→工作目录：gitpull<br><br>开分支：git checkout -b experiment<br><br>删除不需要的或者临时文件：rm+文件名或rm -rf 文件夹（-rf中，r指的是删除当前文件夹中的内容后删除文件夹，f指的是强制删除，比如说源文件夹中文件如果是只读的话，不加f会进行删除询问，加上之后直接删除。ps：临时文件和文件夹可以直接rm或rm -rf直接删除，但是正常/源文件代码最好不要，因为如果用的话是不会放入回收站撤销的）<br><br>保存本地仓库并注释：git commit -m "注释内容"<br><br>

# 为实验开分支

git checkout -b experiment/new-optimizer	-b是创建并切换到一个名为 experiment/new-optimizer 的新分支。这相当于从当前主分支（main）复制了一份完全独立的副本，你之后所有的代码修改都只在这个副本里进行。<br>
……改代码，提交……	在 experiment/new-optimizer 分支上，你可以随意折腾（修改代码、添加文件、提交）。这些提交只影响当前分支，完全不会污染 main 分支。<br>
git checkout main	切换回主分支 main。这时候你的代码会回到没改之前的状态，因为你在 experiment 里做的改动还没合并过来。<br>
git merge experiment/new-optimizer	把实验分支上的成果合并到主分支。Git 会把你在 experiment 分支上新增/修改的所有内容，应用到当前的 main 分支上。如果实验成功，这个操作就把“平行宇宙”里的成果拉回了现实。<br>
git branch -d experiment/new-optimizer
-d 是安全删除，如果分支还没合并，Git 会提醒你，防止丢失未合并的代码。
如果你想强制删除（比如实验失败了，确定不要了），可以用 -D。
