# 为实验开分支

git checkout -b experiment/new-optimizer	-b是创建并切换到一个名为 experiment/new-optimizer 的新分支。这相当于从当前主分支（main）复制了一份完全独立的副本，你之后所有的代码修改都只在这个副本里进行。<br>
git push -u origin+分支名称  把本地的分支第一次推送到 GitHub 云端，并建立本地和远程的“跟踪关系”。<br>
……改代码，提交……	在 experiment/new-optimizer 分支上，你可以随意折腾（修改代码、添加文件、提交）。这些提交只影响当前分支，完全不会污染 main 分支。<br>
git checkout main	切换回主分支 main。这时候你的代码会回到没改之前的状态，因为你在 experiment 里做的改动还没合并过来。<br>
git merge experiment/new-optimizer	把实验分支上的成果合并到主分支。Git 会把你在 experiment 分支上新增/修改的所有内容，应用到当前的 main 分支上。如果实验成功，这个操作就把“平行宇宙”里的成果拉回了现实。然后继续将本地合并的传到远端：git push origin main<br>
git branch -d experiment/new-optimizer
-d 是安全删除，如果分支还没合并，Git 会提醒你，防止丢失未合并的代码。如果你想强制删除（比如实验失败了，确定不要了），可以用 -D。<br>
删除远程分支（如果你确定不再需要它）
git push origin --delete experiment/bew-optimizer<br>
结束后可git status验证

# .gitignore文件的作用

在任何当前工作的 Git 仓库中，每个文件都是这样的：<br>
追踪的（tracked）- 这些是 Git 所知道的所有文件或目录。这些是新添加（用 git add 添加）和提交（用 git commit 提交）到主仓库的文件和目录。<br>
未被追踪的（untracked） - 这些是在工作目录中创建的，但还没有被暂存（或用 git add 命令添加）的任何新文件或目录。<br>
被忽略的（ignored） - 这些是 Git 知道的要全部排除、忽略或在 Git 仓库中不需要注意的所有文件或目录。本质上，这是一种告诉 Git 哪些未被追踪的文件应该保持不被追踪并且永远不会被提交的方法。<br>
所有被忽略的文件都会被保存在一个 .gitignore 文件中。<br>
.gitignore 文件是一个纯文本文件，包含了项目中所有指定的文件和文件夹的列表，这些文件和文件夹是 Git 应该忽略和不追踪的。<br>
在 .gitignore 中，你可以通过提及特定文件或文件夹的名称或模式来告诉 Git 只忽略一个文件或一个文件夹。你也可以用同样的方法告诉 Git 忽略多个文件或文件夹。<br>
我们主要用来：<br>
1.防止仓库“爆炸”（存储与带宽）<br>
一个 PyTorch 模型权重文件（.pt 或 .safetensors）动辄 几百 MB 甚至几十 GB。<br>
GitHub 严格限制单文件不超过 100 MB，仓库总大小建议不超过 5 GB。如果你不小心把 3 个模型权重文件提交上去，你的仓库会立刻被 GitHub 封禁推送（报错 File exceeds 100 MB），整个项目就瘫痪了。<br>
2.避免无效的 Git 历史（性能）<br>
Git 的设计是为文本代码服务的（差量存储）。对于二进制大文件（权重文件），每次修改哪怕 1 个字节，Git 都会完整存储一份新副本。这会让 .git 文件夹迅速膨胀到几十 GB，导致 git clone 慢如蜗牛。<br>
3.区分“代码”与“数据”<br>
源码（Code）：需要版本控制，因为它是逻辑和智慧，体积小。<br>
权重（Weights）：属于“产物”或“数据依赖”，通常由训练脚本生成，或者从 Hugging Face 下载。你不需要备份它们，因为它们是可复现的（只要有代码、随机种子和数据集）。
