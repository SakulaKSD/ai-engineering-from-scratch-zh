# 神经网络的前向传播

一个神经元做的事情是：

$$y = \sigma(W x+ b)$$

其中：
- $W$ 是权重向量
- $x$ 是输入向量
- $b$ 是偏置
- $\sigma$ 是激活函数，例如 $\sigma(z) = \frac{1}{1 + e^{-z}}$

梯度下降更新规则：

$$\theta \leftarrow w - \alpha \frac{\partial L}{\partial w}$$
