# 7..gitignore文件的作用

在任何当前工作的 Git 仓库中，每个文件都是这样的：<br>
追踪的（tracked）- 这些是 Git 所知道的所有文件或目录。这些是新添加（用 git add 添加）和提交（用 git commit 提交）到主仓库的文件和目录。<br>
未被追踪的（untracked） - 这些是在工作目录中创建的，但还没有被暂存（或用 git add 命令添加）的任何新文件或目录。<br>
被忽略的（ignored） - 这些是 Git 知道的要全部排除、忽略或在 Git 仓库中不需要注意的所有文件或目录。本质上，这是一种告诉 Git 哪些未被追踪的文件应该保持不被追踪并且永远不会被提交的方法。<br>
所有被忽略的文件都会被保存在一个 .gitignore 文件中。<br>
.gitignore 文件是一个纯文本文件，包含了项目中所有指定的文件和文件夹的列表，这些文件和文件夹是 Git 应该忽略和不追踪的。<br>
在 .gitignore 中，你可以通过提及特定文件或文件夹的名称或模式来告诉 Git 只忽略一个文件或一个文件夹。你也可以用同样的方法告诉 Git 忽略多个文件或文件夹。<br>
我们主要用来：<br>
1.防止仓库“爆炸”（存储与带宽）<br>
一个 PyTorch 模型权重文件（.pt 或 .safetensors）动辄 几百 MB 甚至几十 GB。<br>
GitHub 严格限制单文件不超过 100 MB，仓库总大小建议不超过 5 GB。如果你不小心把 3 个模型权重文件提交上去，你的仓库会立刻被 GitHub 封禁推送（报错 File exceeds 100 MB），整个项目就瘫痪了。<br>
2.避免无效的 Git 历史（性能）<br>
Git 的设计是为文本代码服务的（差量存储）。对于二进制大文件（权重文件），每次修改哪怕 1 个字节，Git 都会完整存储一份新副本。这会让 .git 文件夹迅速膨胀到几十 GB，导致 git clone 慢如蜗牛。<br>
3.区分“代码”与“数据”<br>
源码（Code）：需要版本控制，因为它是逻辑和智慧，体积小。<br>
权重（Weights）：属于“产物”或“数据依赖”，通常由训练脚本生成，或者从 Hugging Face 下载。你不需要备份它们，因为它们是可复现的（只要有代码、随机种子和数据集）。

# 8.GPU与CPU对比设置

````python
# 在 WSL 终端中执行，查看 NVIDIA 显卡的实时运行状态
# 输出包括：驱动版本、CUDA 版本、GPU 利用率、显存占用、温度、功耗等
# 常用选项：
#   nvidia-smi          # 默认每隔几秒刷新一次，按 Ctrl+C 退出
#   nvidia-smi --query-gpu=... --format=csv  # 获取特定字段（如显存占用）
#   watch -n 1 nvidia-smi  # 每秒刷新一次（需安装 watch）
nvidia-smi
import torch

# 检查 CUDA（即 NVIDIA GPU 加速）是否可用
# 如果返回 True，说明你的显卡已被 PyTorch 识别
print(f"CUDA available: {torch.cuda.is_available()}")

# 打印 PyTorch 编译时使用的 CUDA 版本（不是驱动版本）
print(f"CUDA version: {torch.version.cuda}")

if torch.cuda.is_available():
    # 获取当前 GPU 的设备名称（例如 "NVIDIA GeForce RTX 5070 Laptop GPU"）
    print(f"GPU: {torch.cuda.get_device_name(0)}")

    # 获取 GPU 总显存，单位转换为 GB（1e9 = 10^9 bytes）
    # total_memory 返回的是字节数，除以 1e9 得到 GB
    mem_gb = torch.cuda.get_device_properties(0).total_memory / 1e9
    print(f"Memory: {mem_gb:.1f} GB")

import torch
import time

# 矩阵大小：5000 x 5000，计算量约为 5000^3 = 1250 亿次乘加
size = 5000
print(f"Matrix size: {size}x{size}")

# ----- CPU 计算（默认在 CPU 上） -----
# torch.randn 生成服从标准正态分布的随机张量
a_cpu = torch.randn(size, size)
b_cpu = torch.randn(size, size)

start = time.time()
# @ 运算符表示矩阵乘法（等价于 torch.matmul）
c_cpu = a_cpu @ b_cpu
cpu_time = time.time() - start
print(f"CPU: {cpu_time:.3f}s")

# ----- GPU 计算（如果 CUDA 可用） -----
if torch.cuda.is_available():
    # .to("cuda") 将张量从 CPU 内存复制到 GPU 显存
    a_gpu = a_cpu.to("cuda")
    b_gpu = b_cpu.to("cuda")

    # torch.cuda.synchronize() 强制等待所有 GPU 操作完成
    # 确保后续计时的起始点准确（防止异步执行干扰）
    torch.cuda.synchronize()
    start = time.time()

    # 矩阵乘法在 GPU 上执行（自动利用 CUDA 内核）
    c_gpu = a_gpu @ b_gpu

    # 再次同步，确保矩阵乘法完全结束，避免计时过早
    torch.cuda.synchronize()
    gpu_time = time.time() - start

    print(f"GPU: {gpu_time:.3f}s")
    # 计算加速比（CPU 耗时 / GPU 耗时），越大说明 GPU 优势越明显
    print(f"Speedup: {cpu_time / gpu_time:.0f}x")
else:
    print("CUDA not available, skipping GPU test")
````

# 9.什么是 Linux Kernel？
Linux Kernel 是 Linux 操作系统的核心组件，负责在计算机硬件与运行的进程之间提供接口。它管理硬件资源并确保系统的高效运行。内核的作用类似于一个“调解者”，在硬件和用户空间的应用程序之间传递信息和请求。

Linux Kernel 的主要功能

内存管理：内核负责跟踪内存的分配和使用情况，包括存储位置和内容。它还管理内存的分页和分段，确保高效的内存使用。

进程管理：内核决定哪些进程可以使用 CPU、使用时长以及调度顺序。它还负责进程的创建、终止和同步，避免死锁。

设备驱动程序：内核通过设备驱动程序与硬件交互，充当硬件与进程之间的翻译器。例如，磁盘控制器、网络适配器等都通过内核进行管理。

系统调用和安全防护：内核提供系统调用接口（SCI），允许用户空间的应用程序请求服务，同时确保系统的安全性和稳定性。

二、重要概念
1. Cell（单元格）：笔记本里可独立运行的基本单元，只有两种——代码（会执行出结果）和 Markdown（渲染成排版好的文档）。
2. Kernel（内核）：后台运行的一个 Python 进程；所有单元格共享它，所以变量在格子间保留——重启内核 = 清空所有变量。
3. 执行顺序 ≠ 格子顺序：能跑通取决于"这次会话按了什么顺序执行"，而不是格子长什么样。这是 Notebook 最大陷阱。
4. 隐藏状态：删掉某个格子后，它创建的变量还留在内存里，容易让人误以为"没依赖它"——要定期重启内核排查。
5. Restart & Run All（可复现性）：分享/交付前，用"重启并运行全部"从头验证一遍，保证别人按顺序跑也能通。
6. 向量化（NumPy 为什么快）：Python 慢在解释开销×循环次数；NumPy 把循环"下沉"到 C 层，Python 只发起一次调用。跟你 MATLAB 里 x.^2 快于 for 循环是同一原理。
7. 工作流铁律：在 notebook 里探索，用脚本交付——跑通的代码要搬进 .py 文件，再 import 回来。

三、实验记录
1. 乱序执行实验：先跑"定义 y"再跑"打印 y"，第 7 格能打印出 5；但 Restart & Run All 就会报 NameError: name 'y' is not defined——亲身体会"执行顺序≠格子顺序"。
2. %timeit 对比：列表推导式（[x**2 for x in range(100000)]）是 ms 级；NumPy（np.arange(100000)**2）是 µs 级，快上千倍。

二、重要概念
1. 虚拟环境：一个独立目录（.venv），内含自己的 Python 解释器和包，与系统 Python 隔离，避免依赖冲突。
2. 依赖地狱：不同项目需要不同（甚至冲突）的包版本，全局安装必崩；虚拟环境是解药。
3. ae / deactivate：进入/退出虚拟环境的开关；提示符前的 (环境名) 标记说明你在环境内。
4. pyproject.toml：项目"菜单"，一个文件替代 setup.py/setup.cfg/requirements.txt，声明直接依赖 + 可选依赖组。
5. 可选依赖组：同一环境里"按需装"的抽屉（如 [torch]、[llm]），不是用来解决版本冲突；不同版本冲突要用不同虚拟环境。
6. uv.lock（lockfile）：把每个包（含传递依赖）钉到精确版本的"采购单"，保证可复现；提交 git，不提交 .venv。
7. 传递依赖：依赖的依赖（装 A 需要 B，B 需要 C，C 就是 A 的传递依赖）。
8. CUDA 不匹配：PyTorch 的 CUDA 版本必须 ≤ 驱动支持的 CUDA 版本，否则 GPU 训练会神秘报错。
9. 全局 vs 虚拟环境：which python 检查包装哪了；"import 不到包"先查 which python（80% 是环境问题）。

三、本节课踩坑记录
1. which python 没输出：不是环境坏了，是"没进虚拟环境 → 系统无全局 python"，ae 后即恢复。
2. uv lock 报错：requires-python = ">=3.11" 太宽，numpy>=2.5 只支持 3.12+，改成 >=3.12 解决（学会看 hint 提示）。
3. python -version 报 Unknown option: -e：单横线 -v 是 verbose，双横线 --version 才是版本。

# 12.面向AI的Docker
Docker 把代码、运行时、库、系统工具打包成"镜像"，在任何机器上都能跑出一样的环境——"在我机器上能跑"成为过去式。本课理解为主：镜像/容器/卷/Compose。

一、终端及代码操作（含解释）——本课以理解为主，命令仅作了解
1. 构建镜像

docker build -t ai-dev -f Dockerfile .：-t 给镜像起名 ai-dev，-f 指定 Dockerfile，. 是构建上下文（当前目录）。

2. 运行容器（带 GPU + 卷挂载）

docker run --rm -it --gpus all \
    -v $(pwd):/workspace \
    -v ~/models:/models \
    ai-dev python -c "import torch; print(torch.cuda.is_available())"

--rm 退出即删容器；-it 交互模式；--gpus all 把 GPU 暴露进容器；-v 宿主机路径:容器路径 是卷挂载（共享文件夹）。

3. 常用命令

docker ps                 # 列出运行中的容器\
docker images             # 列出所有镜像及大小\
docker system prune -a    # 清理未使用镜像（回收磁盘）\
docker compose up -d      # 按 docker-compose.yml 启动全部服务\
docker compose down       # 停止全部服务

二、重要概念

1. 镜像（Image）：只读模板/配方，从 Dockerfile 构建而来；好比 C# 里的"类"。
2. 容器（Container）：镜像运行起来的实例；好比"对象"。共享宿主机内核，秒级启动。
3. Dockerfile：构建镜像的"装配说明书"，每条指令生成一层（RUN 构建时执行、CMD 启动时执行）。
4. 卷（Volume）：宿主机目录映射进容器（-v），容器重建后数据仍在——解决"模型大、不能重下"。
5. GPU 透传：--gpus all 把宿主机 GPU 暴露给容器；容器内 CUDA 版本仍需 ≤ 驱动支持的 CUDA。
6. Docker Compose：用 YAML 定义多容器应用（如开发容器 + 向量数据库），一条命令全启动。
7. 服务发现：Compose 共享网络里，容器间用"服务名"互访（如 host="qdrant"），像内置电话簿；宿主机访问才用端口映射 -p。
8. 基础镜像：一个 预装好操作系统 + 工具包的现成环境（精装房）。FROM 那行：devel 含编译器（构建用）、runtime 只运行、pytorch/pytorch 预装 PyTorch、python:3.12-slim 纯 CPU。
9. 为什么 AI 需要 Docker：GPU 驱动脆弱（CUDA 版本隔离）、模型权重巨大（卷挂载）、多服务架构（Compose）。
10. WSL2 用户跳过 NVIDIA Container Toolkit：Docker Desktop 在 Windows/WSL2 下已自行处理 GPU 透传，无需手动装 toolkit。

三、本课要点记录

1. 课程练习 1/2/3/4 均需真装 Docker 才能实操，本课暂不执行（本地已有完整 uv 环境，课程前中期用不到容器化）。等阶段 13+（Agent/部署）需要时再回来补。
2. 镜像 vs 容器 = 模板 vs 实例；RUN vs CMD = 构建时 vs 启动时。
3. 端口映射 -p 8888:8888 用于"宿主机/浏览器 → 容器"；服务名用于"容器 → 容器"。

**二、重要概念**

1. **配置名（config）**：一个仓库里可以装多个子集（GLUE 里有 mrpc/sst2…；维基有 `20231101.en` 等快照名），`load_dataset` 第二参数就是选哪个子集。
2. **流式数据集**：返回迭代器（类似 Java `Iterator`），一条条取，不能 `[0]` 随机下标访问。价值不在快，在"几百 GB 不用整包下载、内存恒定"。
3. **缓存**：数据集与模型默认缓存在 `~/.cache/huggingface/`；已永久配置 `HF_ENDPOINT=https://hf-mirror.com` 走镜像。
4. **Parquet 为什么小**：列式存储，同列同类型数据排在一起，压缩效率远高于逐行文本（CSV/JSON 里每条记录还重复一遍字段名）。
5. **划分纪律**：train 学习、val 调参、test 只评估一次。混在一起 = 拿考卷做训练，分数虚高（数据泄漏是新手翻车第一大死因）。
6. **`seed=42`**：固定随机种子保证每次切分结果一致——可复现是对比实验的前提。

## 15. Phase 0-10 终端与 Shell（保姆级操作手册）

**本节主题**：掌握 AI 工程师每天在终端里干的 80% 的事——管道筛日志、tmux 守训练、alias 省手指、出错怎么修。

---

**一、管道与重定向（Step 2）**

| 命令 | 干什么 | 记法 |
|---|---|---|
| `cat 文件 \| grep 词 \| wc -l` | 数含某词的行数 | 三根管子接起来 |
| `grep 词 文件 \| awk '{print $NF}'` | 取每行最后**一列** | `$NF`=最后一列 |
| `命令 > 文件` | **覆盖**写入 | 大头朝文件 |
| `命令 >> 文件` | **追加**写入 | 两个大头=不覆盖 |
| `命令 2> 文件` | 存**错误**输出 | `2`=stderr |
| `命令 > 文件 2>&1` | 正常+错误都存同一文件 | `&1`="合并到 stdout" |
| `tail -f 文件` | 实时盯日志 | 训练时灵魂命令 |
| `head -3 文件` / `tail -3` | 看头/看尾 | 排查第一斧 |

**关键理解**：程序天生有三根管——stdin(0)、stdout(1)、stderr(2)；默认都连屏幕/键盘；`|` 把左嘴接右耳。

**失误记录**：`sort -n` 是**从小到大**（升序），要大到小用 `sort -nr`。`>` 和 `>>` 差一条横线，事故差一千行。

---

**二、tmux（Step 4 保姆级）**

**概念**：终端套终端。它跑在**系统层**，你 VS Code 关了、SSH 断了，tmux 里面的程序**原样活着**。AI 训练几小时起步，没有 tmux 等于抱着定时炸弹睡觉。

**结构（三层，别混）**：
- **session（会话）**= 一间工作室（`tmux new -s train` 开一间）
- **window/pane** = 同一屏切成多块，每块跑一个命令
- **prefix `Ctrl+B`** = 所有 tmux 命令的"敲门键"（先敲门，再下令；**不是组合键**！）

**全流程操作表（含中间走的弯路）**：

```bash
# 准备：先激活虚拟环境
ae                            # 一定在外面激活！tmux 继承母体环境
tmux kill-server 2>/dev/null  # 清空所有旧会话（2>/dev/null 是不想看见报错）
tmux ls                       # 应该显示 no server running

# 开工
tmux new -s train             # 开新会话 train，屏幕底部出现绿条

# 在左窗格跑一个动的（看时间每秒跳）
watch -n1 date

# 竖切一刀（Ctrl+B 松手，再 Shift+%）
# 在右窗格跑 htop
htop                          # F10 或 q 退出
# 横切一刀（Ctrl+B 松手，再 Shift+'）
# 在下窗格跑 python 时钟
python -c "import time; [print('tick', i) or time.sleep(2) for i in range(100)]"

# 在窗格间跳（Ctrl+B 松手，再方向键 ←→↑↓）
# 收口袋
Ctrl+B d                       # detach：绿条消失，回到普通终端，显示 [detached ...]

# 看它还在不在
tmux ls                        # train: 1 windows (3 panes) ...

# 钻回去（它一秒没停）
tmux attach -t train

# 清场（两种走法）
# 走法A（挨窗格退）：Ctrl+B d → tmux kill-session -t train
# 走法B（窗格里退）：给 watch 按 Ctrl+C 停程序 → exit 回车关窗格；给 htop 按 F10 → exit；全退完会话自动死
```

**高频易错（全是今天踩的）**：
1. **`Ctrl+B` 不是组合键**：按住敲一下 B，**全松手**，再按命令键。按成 `Ctrl+B+"` 一样同按是没反应的。
2. **detach 后环境还在**：`which python` 进 tmux 后依然指向 `.venv/bin/python` 就对了；**前缀消失（没有 `(venv)`）是 tmux 3.x 故意的**，不影响真身。
3. **退出 tmux 顺序**：先退出窗格里的程序（watch 用 Ctrl+C、htop 用 F10/q），出现 `$` 提示符后才有 `exit`；只剩 watch 卡住时记得先 Ctrl+C。
4. **`ae` 要在 tmux 外面敲**：tmux 从当前 shell 分叉继承；如果你进 tmux 才想起 `ae`，窗格里再敲一次也行，但肌肉记忆应该是 **`ae` 永远第一个**。

---

**三、后台进程（Step 3）**

| 命令 | 效果 | 用法 |
|---|---|---|
| `命令 &` | 丢后台跑 | 终端不被占，但关终端还是死 |
| `nohup 命令 > log 2>&1 &` | 后台 + 防挂断 + 输出进文件 | 租云跑训练的标准式 |
| `jobs` | 看当前终端挂着的后台任务 | `[1]+ Running ...` |
| `fg %1` | 把后台任务拽回前台 | 之后 Ctrl+C/Z 才有效 |
| `bg %1` | 让挂起的任务转后台继续 | Ctrl+Z 挂起后救场 |
| `kill %1` | 处决第一个后台任务 | 比 `pkill` 精准 |

**今天踩的坑**：**`Ctrl+C / Ctrl+Z` 只对前台进程有效**。后台任务要 `jobs` 查工号 → `fg %1` 拽回来再杀，或 `kill %1` 直接处决。
