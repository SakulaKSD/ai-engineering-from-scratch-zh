# 练习 2：从 CSV 加载数据并可视化

本练习完整走一遍「生成数据 → 存 CSV → 读回 → 画图」，
最后用「重启内核并运行全部」验证可复现。

```python
import numpy as np
import pandas as pd

x = np.linspace(0, 10, 50)                       # 0~10 取 50 个点
y = np.sin(x) + np.random.randn(50) * 0.1        # 正弦 + 一点随机噪声

df = pd.DataFrame({"x": x, "y": y})              # 组装成一张表（DataFrame）
df.to_csv("data.csv", index=False)               # 存成 CSV 文件（存在当前目录）

df.head()                                        # 显示前 5 行，确认长啥样
```

```python
df = pd.read_csv("data.csv")   # 从文件读回来，证明数据真的落地了
df.describe()                  # 统计摘要（均值、标准差、最大最小值）
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Noto Sans CJK SC']  # 中文（复习上次修的字体）
plt.rcParams['axes.unicode_minus'] = False

plt.plot(df["x"], df["y"], "o-", label="带噪声的 sin")
plt.xlabel("x")
plt.ylabel("y")
plt.title("从 CSV 读回的数据可视化")
plt.legend()
plt.show()
```
